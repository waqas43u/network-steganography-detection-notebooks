```python
#Covert Techniques Wise distribution

#Header Manipulation
#Timing Obfuscation
# Flow Blending



# ── Define features grouped by technique ───────────────────────────────────

header_manipulation_features = [
    "total_bytes", "receiving_bytes", "sending_bytes",
    "min_packets_len", "max_packets_len",
    "dns_domain_name", "dns_top_level_domain", "dns_second_level_domain",
    "dns_domain_name_length", "dns_subdomain_name_length",
    "uni_gram_domain_name", "bi_gram_domain_name", "tri_gram_domain_name",
    "numerical_percentage", "character_distribution", "character_entropy",
    "max_continuous_numeric_len", "max_continuous_alphabet_len",
    "max_continuous_consonants_len", "max_continuous_same_alphabet_len",
    "vowels_consonant_ratio", "conv_freq_vowels_consonants",
    "ans_resource_record_type", "ans_resource_record_class"
]

time_obfuscation_features = [
    "duration", "distinct_ttl_values",
    "ttl_values_min", "ttl_values_max", "ttl_values_mean",
    "ttl_values_mode", "ttl_values_median"
]

flow_blending_features = [
    "packets_rate", "packets_len_rate", "mean_packets_len",
    "standard_deviation_packets_len", "variance_packets_len",
    "coefficient_of_variation_packets_len", "distinct_A_records"
]

# All flow-level features combined
all_flow_features = (
    header_manipulation_features +
    time_obfuscation_features +
    flow_blending_features
)

# ── 3. Validate: check which features are actually present in the dataset ──────
available     = [f for f in all_flow_features if f in dns_tunneling_df.columns]
missing       = [f for f in all_flow_features if f not in dns_tunneling_df.columns]

print(f"✅ Features found    : {len(available)}/{len(all_flow_features)}")
print(f"❌ Features missing  : {missing}\n")
```

```python
# ── Combine all three technique subsets into one file ──────────────────────────

# Extract each technique's features (only those available in the dataset)
avail_header = [f for f in header_manipulation_features if f in dns_tunneling_df.columns]
avail_time   = [f for f in time_obfuscation_features    if f in dns_tunneling_df.columns]
avail_flow   = [f for f in flow_blending_features       if f in dns_tunneling_df.columns]

# Detect label column
label_col = next((c for c in ["label", "Label", "class", "target"] if c in dns_tunneling_df.columns), None)

# Build each technique subset and tag it
def make_subset(df, features, technique_name, label_col):
    cols = features + ([label_col] if label_col else [])
    subset = df[cols].copy()
    subset["technique"] = technique_name
    return subset

df_header = make_subset(dns_tunneling_df, avail_header, "Header Manipulation", label_col)
df_time   = make_subset(dns_tunneling_df, avail_time,   "Time Obfuscation",    label_col)
df_flow   = make_subset(dns_tunneling_df, avail_flow,   "Flow Blending",       label_col)

# Merge all three into one file
df_combined = pd.concat([df_header, df_time, df_flow], ignore_index=True)

# Save
output_path = "covert_dns_tunneling_all_techniques.csv"
df_combined.to_csv(output_path, index=False)

print(f"✅ Combined dataset saved → {output_path}")
print(f"   Total rows     : {len(df_combined):,}")
print(f"   Total columns  : {df_combined.shape[1]}")
print(f"\n📊 Rows per technique:")
print(df_combined["technique"].value_counts().to_string())
print(f"\n📋 Column preview:\n{df_combined.head(3)}")
```

```python
# ── Sample 3000 flows per technique (label = 1 for all) ───────────────────────

df_sampled = (
    df_combined
    .groupby("technique", group_keys=False)
    .apply(lambda x: x.sample(n=3000, random_state=42))
    .reset_index(drop=True)
)

# Set label = 1 for all (covert/tunneling traffic)
df_sampled["label"] = 1

# Save
df_sampled.to_csv("covert_dns_tunneling_3000_per_technique.csv", index=False)

print(f"✅ Saved → covert_dns_tunneling_3000_per_technique.csv")
print(f"   Total rows : {len(df_sampled):,}")
print(f"\n📊 Rows per technique:")
print(df_sampled["technique"].value_counts().to_string())
print(f"\n🏷️  Label distribution:\n{df_sampled['label'].value_counts().to_string()}")
```

```python
"""
DNS Tunneling — Comparative Visualizations (3000 flows × 3 techniques)
======================================================================
Reads:  covert_dns_tunneling_3000_per_technique.csv

Plots:
  1.  Technique distribution (bar + pie)
  2.  Boxplots — numeric features per technique (all 3 groups)
  3.  Violin plots — key features per technique
  4.  Mean feature bar chart (top discriminative features)
  5.  Correlation heatmaps (one per technique)
  6.  PCA 2-D scatter
  7.  Feature-group radar / spider chart
  8.  KDE density curves for top features
"""

import os, warnings
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
from matplotlib.lines import Line2D
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

warnings.filterwarnings("ignore")
np.random.seed(42)

OUT = "dns_visualizations"
os.makedirs(OUT, exist_ok=True)

# ── Palette ──────────────────────────────────────────────────────────────────
TECH_NAMES  = ["Header Manipulation", "Time Obfuscation", "Flow Blending"]
COLORS      = ["#E63946", "#457B9D", "#2A9D8F"]
PALETTE     = dict(zip(TECH_NAMES, COLORS))
SHORT       = ["HM", "TO", "FB"]

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "figure.facecolor": "#F8F9FA",
    "axes.facecolor": "#F8F9FA",
    "axes.grid": True,
    "grid.alpha": 0.3,
    "grid.linestyle": "--",
})

# ─────────────────────────────────────────────────────────────────────────────
# DATA
# ─────────────────────────────────────────────────────────────────────────────

HEADER_FEATS = [
    "total_bytes","receiving_bytes","sending_bytes",
    "min_packets_len","max_packets_len",
    "dns_domain_name_length","dns_subdomain_name_length",
    "uni_gram_domain_name","bi_gram_domain_name","tri_gram_domain_name",
    "numerical_percentage","character_distribution","character_entropy",
    "max_continuous_numeric_len","max_continuous_alphabet_len",
    "max_continuous_consonants_len","max_continuous_same_alphabet_len",
    "vowels_consonant_ratio","conv_freq_vowels_consonants",
]
TIME_FEATS = [
    "duration","distinct_ttl_values",
    "ttl_values_min","ttl_values_max","ttl_values_mean",
    "ttl_values_mode","ttl_values_median",
]
FLOW_FEATS = [
    "packets_rate","packets_len_rate","mean_packets_len",
    "standard_deviation_packets_len","variance_packets_len",
    "coefficient_of_variation_packets_len","distinct_A_records",
]
ALL_FEATS = HEADER_FEATS + TIME_FEATS + FLOW_FEATS


def make_synthetic():
    """Generate realistic synthetic data for demo purposes."""
    dfs = []
    for i, tech in enumerate(TECH_NAMES):
        n = 3000
        rng = np.random.default_rng(i * 7)
        row = {"technique": tech, "label": 1}

        if tech == "Header Manipulation":
            row.update({
                "total_bytes":                  rng.normal(1200, 400, n).clip(100),
                "receiving_bytes":              rng.normal(800,  300, n).clip(50),
                "sending_bytes":                rng.normal(400,  200, n).clip(50),
                "min_packets_len":              rng.normal(80,   20,  n).clip(20),
                "max_packets_len":              rng.normal(220,  60,  n).clip(80),
                "dns_domain_name_length":       rng.normal(55,   12,  n).clip(10),
                "dns_subdomain_name_length":    rng.normal(30,   8,   n).clip(1),
                "uni_gram_domain_name":         rng.normal(18,   4,   n).clip(5),
                "bi_gram_domain_name":          rng.normal(14,   3,   n).clip(2),
                "tri_gram_domain_name":         rng.normal(10,   3,   n).clip(1),
                "numerical_percentage":         rng.beta(2, 3, n) * 0.6,
                "character_distribution":       rng.normal(0.5,  0.1, n).clip(0,1),
                "character_entropy":            rng.normal(3.8,  0.5, n).clip(0),
                "max_continuous_numeric_len":   rng.normal(6,    2,   n).clip(1),
                "max_continuous_alphabet_len":  rng.normal(8,    3,   n).clip(1),
                "max_continuous_consonants_len":rng.normal(5,    2,   n).clip(1),
                "max_continuous_same_alphabet_len": rng.normal(2, 1,  n).clip(1),
                "vowels_consonant_ratio":       rng.normal(0.4,  0.1, n).clip(0),
                "conv_freq_vowels_consonants":  rng.normal(0.3,  0.08,n).clip(0),
                # time/flow cols = NaN
                **{f: np.nan for f in TIME_FEATS + FLOW_FEATS},
            })

        elif tech == "Time Obfuscation":
            row.update({
                "duration":             rng.exponential(30, n).clip(0.1),
                "distinct_ttl_values":  rng.poisson(3, n).clip(1),
                "ttl_values_min":       rng.normal(15,  10, n).clip(0),
                "ttl_values_max":       rng.normal(300, 80, n).clip(10),
                "ttl_values_mean":      rng.normal(120, 40, n).clip(5),
                "ttl_values_mode":      rng.normal(60,  30, n).clip(1),
                "ttl_values_median":    rng.normal(90,  35, n).clip(1),
                **{f: np.nan for f in HEADER_FEATS + FLOW_FEATS},
            })

        else:   # Flow Blending
            row.update({
                "packets_rate":                       rng.normal(12,  5,   n).clip(0.1),
                "packets_len_rate":                   rng.normal(800, 300, n).clip(10),
                "mean_packets_len":                   rng.normal(130, 40,  n).clip(20),
                "standard_deviation_packets_len":     rng.normal(45,  15,  n).clip(1),
                "variance_packets_len":               rng.normal(2100,800, n).clip(10),
                "coefficient_of_variation_packets_len": rng.normal(0.35,0.1,n).clip(0),
                "distinct_A_records":                 rng.poisson(4, n).clip(1),
                **{f: np.nan for f in HEADER_FEATS + TIME_FEATS},
            })

        dfs.append(pd.DataFrame(row))

    return pd.concat(dfs, ignore_index=True)


def load_data():
    path = "covert_dns_tunneling_3000_per_technique.csv"
    if os.path.exists(path):
        df = pd.read_csv(path)
        print(f"✅ Loaded real dataset: {df.shape}")
    else:
        print("⚠️  CSV not found → using synthetic demo data")
        df = make_synthetic()
    return df


def numeric_cols(df):
    cols = [c for c in ALL_FEATS if c in df.columns
            and df[c].dtype != object]
    return cols


def save(fig, name, dpi=150):
    p = os.path.join(OUT, name)
    fig.savefig(p, dpi=dpi, bbox_inches="tight")
    plt.close(fig)
    print(f"  💾 {p}")


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 1 — Distribution bar + pie
# ─────────────────────────────────────────────────────────────────────────────
def plot_distribution(df):
    counts = df["technique"].value_counts().reindex(TECH_NAMES)

    fig = plt.figure(figsize=(14, 5), facecolor="#F8F9FA")
    fig.suptitle("DNS Tunneling — Sample Distribution by Technique",
                 fontsize=15, fontweight="bold", y=1.02)

    gs = gridspec.GridSpec(1, 2, figure=fig, wspace=0.35)
    ax1, ax2 = fig.add_subplot(gs[0]), fig.add_subplot(gs[1])

    bars = ax1.bar(SHORT, counts.values,
                   color=COLORS, edgecolor="white", linewidth=1.5, width=0.5)
    for bar, v in zip(bars, counts.values):
        ax1.text(bar.get_x()+bar.get_width()/2, bar.get_height()+30,
                 f"{v:,}", ha="center", fontsize=11, fontweight="bold")
    ax1.set_ylim(0, counts.max() * 1.15)
    ax1.set_ylabel("Number of Flows", fontsize=11)
    ax1.set_title("Sample Count per Technique", fontsize=12)
    ax1.tick_params(axis="x", labelsize=10)
    legend_p = [mpatches.Patch(color=c, label=t) for t,c in zip(TECH_NAMES, COLORS)]
    ax1.legend(handles=legend_p, fontsize=9, frameon=False)

    wedges, texts, autos = ax2.pie(
        counts.values, labels=TECH_NAMES, colors=COLORS,
        autopct="%1.1f%%", startangle=140,
        wedgeprops=dict(edgecolor="white", linewidth=2))
    for at in autos: at.set_fontsize(11)
    ax2.set_title("Proportion of Each Technique", fontsize=12)

    plt.tight_layout()
    save(fig, "01_technique_distribution.png")


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 2 — Boxplots per feature group
# ─────────────────────────────────────────────────────────────────────────────
def plot_boxplots(df, feats, group_name, fname):
    cols = [f for f in feats if f in df.columns and df[f].dtype != object]
    if not cols: return

    ncols = 4
    nrows = int(np.ceil(len(cols) / ncols))
    fig, axes = plt.subplots(nrows, ncols,
                             figsize=(ncols*3.8, nrows*3.4),
                             facecolor="#F8F9FA")
    axes = np.array(axes).flatten()
    fig.suptitle(f"Boxplots — {group_name} Features by Technique",
                 fontsize=14, fontweight="bold")

    for ax, feat in zip(axes, cols):
        data = [df.loc[df["technique"]==t, feat].dropna().values for t in TECH_NAMES]
        bp = ax.boxplot(data, patch_artist=True, notch=False,
                        medianprops=dict(color="black", linewidth=2.5),
                        whiskerprops=dict(linewidth=1.2),
                        capprops=dict(linewidth=1.2),
                        flierprops=dict(marker="o", markersize=2, alpha=0.3))
        for patch, col in zip(bp["boxes"], COLORS):
            patch.set_facecolor(col); patch.set_alpha(0.75)
        ax.set_xticklabels(SHORT, fontsize=9)
        ax.set_title(feat.replace("_"," "), fontsize=8.5, fontweight="bold")

    for ax in axes[len(cols):]: ax.set_visible(False)

    lp = [mpatches.Patch(color=c, label=t, alpha=0.75)
          for t, c in zip(TECH_NAMES, COLORS)]
    fig.legend(handles=lp, loc="lower center", ncol=3,
               fontsize=10, frameon=False, bbox_to_anchor=(0.5, -0.02))
    plt.tight_layout()
    save(fig, fname)


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 3 — Violin plots
# ─────────────────────────────────────────────────────────────────────────────
def plot_violins(df, feats, group_name, fname, max_f=6):
    cols = [f for f in feats if f in df.columns and df[f].dtype != object][:max_f]
    if not cols: return

    ncols = 3
    nrows = int(np.ceil(len(cols) / ncols))
    fig, axes = plt.subplots(nrows, ncols,
                             figsize=(ncols*4.5, nrows*4),
                             facecolor="#F8F9FA")
    axes = np.array(axes).flatten()
    fig.suptitle(f"Violin Plots — {group_name} Features by Technique",
                 fontsize=14, fontweight="bold")

    for ax, feat in zip(axes, cols):
        sub = df[["technique", feat]].dropna()
        sns.violinplot(data=sub, x="technique", y=feat,
                       palette=PALETTE, order=TECH_NAMES,
                       inner="quartile", linewidth=1.2, ax=ax)
        ax.set_xticklabels(SHORT, fontsize=9)
        ax.set_title(feat.replace("_"," "), fontsize=9, fontweight="bold")
        ax.set_xlabel("")

    for ax in axes[len(cols):]: ax.set_visible(False)

    lp = [mpatches.Patch(color=PALETTE[t], label=t) for t in TECH_NAMES]
    fig.legend(handles=lp, loc="lower center", ncol=3,
               fontsize=10, frameon=False, bbox_to_anchor=(0.5, -0.02))
    plt.tight_layout()
    save(fig, fname)


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 4 — Mean feature bar chart (top discriminative)
# ─────────────────────────────────────────────────────────────────────────────
def plot_mean_bars(df, num_cols):
    scaler = StandardScaler()
    filled = df[num_cols].fillna(df[num_cols].median())
    scaled = scaler.fit_transform(filled)
    df_s = pd.DataFrame(scaled, columns=num_cols)
    df_s["technique"] = df["technique"].values

    means = df_s.groupby("technique")[num_cols].mean()
    top_feats = means.std(axis=0).nlargest(15).index.tolist()
    means_top = means[top_feats]

    fig, ax = plt.subplots(figsize=(16, 6), facecolor="#F8F9FA")
    x = np.arange(len(top_feats))
    w = 0.26

    for i, (tech, color) in enumerate(zip(TECH_NAMES, COLORS)):
        ax.bar(x + i*w, means_top.loc[tech], w,
               label=tech, color=color, alpha=0.85, edgecolor="white",
               linewidth=0.8)

    ax.set_xticks(x + w)
    ax.set_xticklabels([f.replace("_","\n") for f in top_feats],
                       fontsize=8, rotation=30, ha="right")
    ax.set_ylabel("Standardised Mean Value", fontsize=11)
    ax.set_title("Top 15 Most Discriminative Features — Standardised Mean by Technique",
                 fontsize=13, fontweight="bold")
    ax.axhline(0, color="#555", linewidth=0.8, linestyle="--")
    ax.legend(frameon=False, fontsize=10)
    plt.tight_layout()
    save(fig, "04_mean_feature_barchart.png")


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 5 — Correlation heatmaps (one per technique)
# ─────────────────────────────────────────────────────────────────────────────
def plot_heatmaps(df, num_cols):
    fig, axes = plt.subplots(1, 3, figsize=(24, 9), facecolor="#F8F9FA")
    fig.suptitle("Feature Correlation Heatmaps by Technique",
                 fontsize=14, fontweight="bold")

    for ax, tech, col in zip(axes, TECH_NAMES, COLORS):
        sub = df.loc[df["technique"]==tech, num_cols].dropna(axis=1, how="all")
        sub = sub.dropna()
        if sub.empty or sub.shape[1] < 2:
            ax.set_title(f"{tech}\n(insufficient data)", color=col)
            continue
        corr = sub.corr()
        mask = np.triu(np.ones_like(corr, dtype=bool))
        sns.heatmap(corr, ax=ax, mask=mask, cmap="coolwarm",
                    center=0, vmin=-1, vmax=1, square=True,
                    linewidths=0.3, cbar_kws={"shrink": 0.75},
                    xticklabels=[c[:12] for c in corr.columns],
                    yticklabels=[c[:12] for c in corr.columns])
        ax.set_title(tech, fontsize=12, fontweight="bold", color=col, pad=10)
        ax.tick_params(labelsize=7)

    plt.tight_layout()
    save(fig, "05_correlation_heatmaps.png")


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 6 — PCA 2-D scatter
# ─────────────────────────────────────────────────────────────────────────────
def plot_pca(df, num_cols):
    filled = df[num_cols].fillna(df[num_cols].median())
    X = StandardScaler().fit_transform(filled)
    comps = PCA(n_components=2, random_state=42).fit(X)
    ev = comps.explained_variance_ratio_ * 100
    Z  = comps.transform(X)

    fig, ax = plt.subplots(figsize=(10, 7), facecolor="#F8F9FA")
    for tech, color in zip(TECH_NAMES, COLORS):
        idx = df["technique"].values == tech
        ax.scatter(Z[idx,0], Z[idx,1], c=color, label=tech,
                   alpha=0.35, s=12, edgecolors="none")

    ax.set_xlabel(f"PC1  ({ev[0]:.1f}% variance)", fontsize=11)
    ax.set_ylabel(f"PC2  ({ev[1]:.1f}% variance)", fontsize=11)
    ax.set_title("PCA 2-D Projection — Separability of Three Tunneling Techniques",
                 fontsize=13, fontweight="bold")
    ax.legend(frameon=True, fontsize=10, markerscale=2)
    plt.tight_layout()
    save(fig, "06_pca_scatter.png")


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 7 — Radar / spider chart (per feature group mean)
# ─────────────────────────────────────────────────────────────────────────────
def plot_radar(df):
    # Use top 8 features from each group that exist and are numeric
    groups = {
        "Header Manipulation Features": HEADER_FEATS,
        "Time Obfuscation Features":    TIME_FEATS,
        "Flow Blending Features":       FLOW_FEATS,
    }

    for grp_title, feats in groups.items():
        cols = [f for f in feats
                if f in df.columns and df[f].dtype != object][:8]
        if len(cols) < 3: continue

        filled = df[cols + ["technique"]].copy()
        for c in cols:
            filled[c] = filled[c].fillna(filled[c].median())
        scaler = StandardScaler()
        filled[cols] = scaler.fit_transform(filled[cols])
        means = filled.groupby("technique")[cols].mean()
        # shift to [0,1]
        mn, mx = means.values.min(), means.values.max()
        norm = (means - mn) / (mx - mn + 1e-9)

        N = len(cols)
        angles = np.linspace(0, 2*np.pi, N, endpoint=False).tolist()
        angles += angles[:1]

        fig, ax = plt.subplots(figsize=(8, 8),
                               subplot_kw=dict(polar=True),
                               facecolor="#F8F9FA")
        ax.set_facecolor("#F8F9FA")

        for tech, color in zip(TECH_NAMES, COLORS):
            if tech not in norm.index: continue
            vals = norm.loc[tech].tolist() + [norm.loc[tech].tolist()[0]]
            ax.plot(angles, vals, color=color, linewidth=2, label=tech)
            ax.fill(angles, vals, color=color, alpha=0.12)

        ax.set_xticks(angles[:-1])
        ax.set_xticklabels([c.replace("_","\n") for c in cols],
                           fontsize=8, fontweight="bold")
        ax.set_yticklabels([])
        ax.set_title(f"Radar Chart — {grp_title}",
                     fontsize=12, fontweight="bold", pad=20)
        ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1),
                  fontsize=9, frameon=False)
        ax.grid(color="gray", alpha=0.3)

        fname = f"07_radar_{grp_title.split()[0].lower()}.png"
        plt.tight_layout()
        save(fig, fname)


# ─────────────────────────────────────────────────────────────────────────────
# PLOT 8 — KDE density curves for top discriminative features
# ─────────────────────────────────────────────────────────────────────────────
def plot_kde(df, num_cols):
    scaler = StandardScaler()
    filled = df[num_cols].fillna(df[num_cols].median())
    scaled = scaler.fit_transform(filled)
    df_s   = pd.DataFrame(scaled, columns=num_cols)
    df_s["technique"] = df["technique"].values

    means    = df_s.groupby("technique")[num_cols].mean()
    top_feats = means.std(axis=0).nlargest(9).index.tolist()

    ncols, nrows = 3, 3
    fig, axes = plt.subplots(nrows, ncols,
                             figsize=(ncols*4.5, nrows*3.5),
                             facecolor="#F8F9FA")
    axes = axes.flatten()
    fig.suptitle("KDE Density Curves — Top 9 Discriminative Features by Technique",
                 fontsize=14, fontweight="bold")

    for ax, feat in zip(axes, top_feats):
        for tech, color in zip(TECH_NAMES, COLORS):
            data = df_s.loc[df_s["technique"]==tech, feat].dropna()
            if len(data) < 10: continue
            sns.kdeplot(data, ax=ax, color=color, linewidth=2,
                        fill=True, alpha=0.15, label=tech)
        ax.set_title(feat.replace("_"," "), fontsize=9, fontweight="bold")
        ax.set_xlabel("Standardised Value", fontsize=8)
        ax.set_ylabel("Density", fontsize=8)

    for ax in axes[len(top_feats):]: ax.set_visible(False)

    lp = [Line2D([0],[0], color=c, linewidth=2.5, label=t)
          for t, c in zip(TECH_NAMES, COLORS)]
    fig.legend(handles=lp, loc="lower center", ncol=3,
               fontsize=10, frameon=False, bbox_to_anchor=(0.5, -0.02))
    plt.tight_layout()
    save(fig, "08_kde_density_curves.png")


# ─────────────────────────────────────────────────────────────────────────────
# MAIN
# ─────────────────────────────────────────────────────────────────────────────
def main():
    print("="*60)
    print(" DNS Tunneling — Visualization Pipeline")
    print("="*60)

    df = load_data()
    num = numeric_cols(df)
    print(f"Numeric features available: {len(num)}\n")

    print("📊 Generating plots…")
    plot_distribution(df)

    plot_boxplots(df, HEADER_FEATS, "Header Manipulation", "02a_boxplots_header.png")
    plot_boxplots(df, TIME_FEATS,   "Time Obfuscation",    "02b_boxplots_time.png")
    plot_boxplots(df, FLOW_FEATS,   "Flow Blending",       "02c_boxplots_flow.png")

    plot_violins(df, HEADER_FEATS, "Header Manipulation",  "03a_violins_header.png")
    plot_violins(df, TIME_FEATS,   "Time Obfuscation",     "03b_violins_time.png")
    plot_violins(df, FLOW_FEATS,   "Flow Blending",        "03c_violins_flow.png")

    plot_mean_bars(df, num)
    plot_heatmaps(df, num)
    plot_pca(df, num)
    plot_radar(df)
    plot_kde(df, num)

    files = sorted(os.listdir(OUT))
    print(f"\n✅ Done — {len(files)} plots saved in '{OUT}/':")
    for f in files:
        print(f"   {f}")

if __name__ == "__main__":
    main()
```
